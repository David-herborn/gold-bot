```python
import yfinance as yf
import plotly.graph_objects as go

# 1. Daten für mehrere Assets (C: Andere Märkte)
tickers = ["GC=F", "SI=F"] # Gold und Silber
data = yf.download(tickers, period="1mo")
# Spaltennamen bereinigen
data.columns = [f"{col[0]}_{col[1]}" for col in data.columns]

# 2. Strategie-Tuning (B: Toleranzschwelle)
# Wir kaufen nur, wenn der Preis den MA10 um mindestens 10 USD übersteigt
def check_signal(price, ma, tolerance=10):
    if price > (ma + tolerance):
        return "KAUFEN"
    elif price < (ma - tolerance):
        return "VERKAUFEN"
    else:
        return "WARTEN (Trend unklar)"

# Analyse für Gold
gold_close = data['Close_GC=F'].iloc[-1]
gold_ma10 = data['Close_GC=F'].rolling(window=10).mean().iloc[-1]
signal = check_signal(gold_close, gold_ma10)

# 3. Automatisierte Benachrichtigung (A: Feedback)
print(f"--- ANALYSEBERICHT ---")
print(f"Gold Preis: {gold_close:.2f} | MA10: {gold_ma10:.2f}")
print(f"BOT-ENTSCHEIDUNG: {signal}")

# Chart (C: Vergleich Silber vs Gold)
fig = go.Figure()
fig.add_trace(go.Scatter(x=data.index, y=data['Close_GC=F'], name='Gold'))
fig.add_trace(go.Scatter(x=data.index, y=data['Close_SI=F'], name='Silber'))
fig.update_layout(title='Marktvergleich: Gold vs Silber')
fig.show()
```

```python
# 1. RSI Funktion definieren
def berechne_rsi(data, window=14):
    delta = data['Close'].diff()
    gain = (delta.where(delta > 0, 0)).rolling(window=window).mean()
    loss = (-delta.where(delta < 0, 0)).rolling(window=window).mean()
    rs = gain / loss
    return 100 - (100 / (1 + rs))

# 2. RSI zum Datensatz hinzufügen
gold_data['RSI'] = berechne_rsi(gold_data)

# 3. Erweiterte Entscheidungslogik
rsi_aktuell = gold_data['RSI'].iloc[-1]

print(f"Aktueller RSI-Wert: {rsi_aktuell:.2f}")

if rsi_aktuell > 70:
    print("RSI-SIGNAL: Überkauft (Vorsicht, Korrektur möglich!)")
elif rsi_aktuell < 30:
    print("RSI-SIGNAL: Überverkauft (Chance auf Anstieg!)")
else:
    print("RSI-SIGNAL: Neutraler Bereich.")
```

```python
import yfinance as yf
import pandas as pd

class GoldTradingBot:
    def __init__(self, ticker="GC=F"):
        self.ticker = ticker
        self.data = None
        self.dollar_data = None

    def fetch_data(self, period="2y"):
        # Daten abrufen
        self.data = yf.download(self.ticker, period=period)
        self.dollar_data = yf.download("UUP", period=period)

        # Falls die Daten-Struktur ein MultiIndex ist (passiert bei yfinance oft), glätten wir sie
        if isinstance(self.data.columns, pd.MultiIndex):
            self.data.columns = self.data.columns.get_level_values(0)
        if isinstance(self.dollar_data.columns, pd.MultiIndex):
            self.dollar_data.columns = self.dollar_data.columns.get_level_values(0)

        print("--- SYSTEM STATUS: DATEN SAUBER GELADEN ---")

    def check_dollar_context(self):
        # Wir greifen auf den Close-Preis zu
        # .iloc[-1] ist der aktuellste Wert, .iloc[-10] ist der von vor 10 Tagen
        last_close = self.dollar_data['Close'].iloc[-1]
        ten_days_ago = self.dollar_data['Close'].iloc[-10]

        # Hier findet der Vergleich statt
        if last_close > ten_days_ago:
            return "VORSICHT: Dollar ist stark, das drückt oft auf Gold."
        else:
            return "GÜNSTIG: Dollar-Trend schwach, gut für Gold."

# Initialisierung und Ausführung
bot = GoldTradingBot()
bot.fetch_data()
ergebnis = bot.check_dollar_context()
print(f"Dollar-Wächter sagt: {ergebnis}")
```

```python
import yfinance as yf
import pandas as pd
import numpy as np
from datetime import datetime

class GoldTradingBotPro:
    def __init__(self, ticker="GC=F"):
        self.ticker = ticker
        self.data = None

    def fetch_and_prepare(self):
        # 1. Daten laden
        self.data = yf.download(self.ticker, period="60d", interval="1h")
        if isinstance(self.data.columns, pd.MultiIndex):
            self.data.columns = self.data.columns.get_level_values(0)

        # 2. Indikatoren berechnen
        self.data['MA'] = self.data['Close'].rolling(window=20).mean()
        self.data['ATR'] = (self.data['High'] - self.data['Low']).rolling(window=14).mean()

        # RSI Berechnung
        delta = self.data['Close'].diff()
        gain = (delta.where(delta > 0, 0)).rolling(window=14).mean()
        loss = (-delta.where(delta < 0, 0)).rolling(window=14).mean()
        rs = gain / loss
        self.data['RSI'] = 100 - (100 / (1 + rs))
        print("--- SYSTEM STATUS: DATEN & RSI BEREIT ---")

    def run_full_analysis(self):
        last_row = self.data.iloc[-1]

        # Logik: Trend (MA) + Momentum (RSI)
        is_trend_up = last_row['Close'] > last_row['MA']
        is_not_overbought = last_row['RSI'] < 70

        if is_trend_up and is_not_overbought:
            return f"SIGNAL: KAUFEN | RSI: {last_row['RSI']:.2f} | Trend bestätigt."
        else:
            return f"WARTEN: Kein Kaufsignal | RSI: {last_row['RSI']:.2f} | Preis unter MA."

# --- AUSFÜHRUNG ---
bot = GoldTradingBotPro()
bot.fetch_and_prepare()
print(bot.run_full_analysis())
```
